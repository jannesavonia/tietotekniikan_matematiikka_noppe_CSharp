### Harjoituksia

Kaikkien tehtävien tarkemmat ohjeet löytyvät koodipohjista.

-   **M1.** Toteuta Caesar-salain.

-   **M2.** Vastaanotit viestin $C$=`"MUUN JCMNYYH NUMNU NYBNUPUMNU"`.
    Pura viesti kaikilla mahdollisilla avaimilla, kuten aiemmassa
    esimerkissä toimittiin. Ohjelman ei tarvitse päättää mikä avain on
    oikea.
