# Lukuteoriaa

## Suurin yhteinen tekijä

<hr class="theorem-block-start" style="border:0; border-top:4px solid #1565c0; margin-top:1.2em; margin-bottom:0.6em;">

#### Definition

Kokonaislukujen $a$ ja $b$ **suurin yhteinen tekijä**
$$d = \operatorname{syt}\left(a,\,b\right)$$ jolle pätee
$$d \vert a \ \textnormal{ ja }\ d \vert b$$ ja jos $c \vert a$ ja
$c \vert b$, niin $c\le d$.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #1565c0; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #00695c; margin-top:1.2em; margin-bottom:0.6em;">

#### Remark

Määritelmä sanoo, että $d$ on suurin luku, joka on sekä $a$:n, että
$b$:n tekijä.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #00695c; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #00695c; margin-top:1.2em; margin-bottom:0.6em;">

#### Remark

Määritelmä ei antanut mitään keinoa laskea $\operatorname{syt}\left(a,\,b\right)$.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #00695c; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #ef6c00; margin-top:1.2em; margin-bottom:0.6em;">

#### Example 5

-   $\operatorname{syt}\left(4,\,6\right) = 2$

-   $\operatorname{syt}\left(4,\,-4\right) = 4$

-   $\operatorname{syt}\left(99,\,13\right) = 1$

-   $\operatorname{syt}\left(26,\,33\right) = 1$

-   $\operatorname{syt}\left(38,\,57\right) = 19$

-   $\operatorname{syt}\left(65537,\,256\right) = 1$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #ef6c00; margin-top:0.6em; margin-bottom:1.2em;">

### Harjoituksia

-   **1.** Mitkä seuraavista luvuista luku 4 jakaa

    1.  $18$

    2.  $36$

    3.  $-4$

-   **2.** Laske $\operatorname{syt}\left(35,\,14\right)$

-   **3.** Laske $\operatorname{syt}\left(102,\,22\right)$

Vastaukset seuraavalla kalvolla.

### Ratkaisuja

-   Mitkä seuraavista luvuista luku 4 jakaa

    1.  $18$ – Ei, koska $18=4 \cdot 4 +2$, eli jakojäännös $18/4$ on
        $2$

    2.  $36$ – Jakaa, koska $36 = 9 \cdot 4$

    3.  $-4$ – Jakaa, koska $-4 = -1 \cdot 4$

-   $\operatorname{syt}\left(35,\,14\right) = 7$

-   $\operatorname{syt}\left(102,\,22\right) = 2$
