# Lukuteoriaa

## Pienin yhteinen jaettava

<hr class="theorem-block-start" style="border:0; border-top:4px solid #1565c0; margin-top:1.2em; margin-bottom:0.6em;">

#### Definition

Kokonaislukujen $a$, $b$ **pienin yhteinen jaettava** $c=\operatorname{pyj}\left(a,\,b\right)$
on pienin luonnollinen luku, jolle $a\vert c$ ja $b \vert c$.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #1565c0; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #6a1b9a; margin-top:1.2em; margin-bottom:0.6em;">

#### Theorem

Olkoon $a$ ja $b$ kokonaislukuja. Tällöin
$$\operatorname{syt}\left(a,\,b\right) = \frac{\lvert ab \rvert}{\operatorname{pyj}\left(a,\,b\right)}$$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #6a1b9a; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #ad1457; margin-top:1.2em; margin-bottom:0.6em;">

#### Corollary

$$\operatorname{pyj}\left(a,\,b\right) = \frac{\lvert ab \rvert}{\operatorname{syt}\left(a,\,b\right)}$$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #ad1457; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #ef6c00; margin-top:1.2em; margin-bottom:0.6em;">

#### Example 7

-   $\operatorname{pyj}\left(2,\,3\right) = 6$

-   $\operatorname{pyj}\left(2,\,4\right) = 4$

-   $\operatorname{pyj}\left(10,\,15\right) =  \frac{\lvert 150 \rvert}{5}  = 30$

-   $\operatorname{pyj}\left(7,\,3\right) =  \frac{\lvert 21 \rvert}{1}  = 21$

-   $\operatorname{pyj}\left(33,\,22\right) =  \frac{\lvert 726 \rvert}{11}  = 66$

-   $\operatorname{pyj}\left(9,\,11\right) =  \frac{\lvert 99 \rvert}{1}  = 99$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #ef6c00; margin-top:0.6em; margin-bottom:1.2em;">
