# Lukuteoriaa

## Eulerin funktio

<hr class="theorem-block-start" style="border:0; border-top:4px solid #1565c0; margin-top:1.2em; margin-bottom:0.6em;">

#### Definition

Luvut $a,b \in \mathbb{Z}$ ovat **suhteellisia alkulukuja**, jos
$\operatorname{syt}\left(a,\,b\right)=1$.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #1565c0; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #ef6c00; margin-top:1.2em; margin-bottom:0.6em;">

#### Example 10

-   $3$ ja $7$ ovat suhteellisia alkulukuja

-   $3$ ja $12$ ei, koska $\operatorname{syt}\left(3,\,12\right)=3$

-   $15$ ja $22$ ovat suhteellisia alkulukuja

-   $33$ ja $121$ ei, koska $\operatorname{syt}\left(33,\,121\right)=11$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #ef6c00; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #1565c0; margin-top:1.2em; margin-bottom:0.6em;">

#### Definition

Olkoon $n \in \mathbb{N}$. Tällöin **Eulerin funktio**
$$\varphi(n) = \#\{ m\ : \ m \le n,\ m \in \mathbb{N}, \ \operatorname{syt}\left(m,\,n\right)=1 \}$$
palauttaa niiden enintään $n$:n suuruisten luonnollisten lukujen
lukumäärän, jotka ovat $n$:n kanssa suhteellisia alkulukuja.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #1565c0; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #ef6c00; margin-top:1.2em; margin-bottom:0.6em;">

#### Example 11

-   $\varphi(10)=4$, koska ainoat luonnolliset luvut $a \le 10$,
    joiden $\operatorname{syt}\left(10,\,a\right)=1$ ovat $1$, $3$, $7$ ja $9$

-   $\varphi(1)=1$, koska $\operatorname{syt}\left(1,\,1\right)=1$

-   $\varphi(16)=8$, koska $\operatorname{syt}\left(16,\,k\right)=1$ vain luvuille $k=15$, $13$,
    $11$, $9$, $7$, $5$, $3$, $1$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #ef6c00; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #6a1b9a; margin-top:1.2em; margin-bottom:0.6em;">

#### Theorem

-   Jos $p \in \mathbb{P}$, niin $\varphi(p)=p-1$

-   Jos $\operatorname{syt}\left(m,\,n\right)=1$, niin $\varphi(mn)=\varphi(m)\varphi(n)$

-   Jos $p,q \in \mathbb{P}$ ja $p\ne q$, niin $\varphi(pq)=(p-1)(q-1)$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #6a1b9a; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #ef6c00; margin-top:1.2em; margin-bottom:0.6em;">

#### Example 12

-   $\varphi(11)=11-1=10$, koska $11 \in \mathbb{P}$

-   $\varphi(60) = \varphi(15 \cdot 4)= \varphi(15)\varphi(4) = 8 \cdot 2 = 16$,
    koska $\operatorname{syt}\left(15,\,4\right)=1$

-   $\varphi(21) = \varphi(3 \cdot 7) = (3-1) \cdot (7-1) = 12$, koska
    $3 \ne 7$ ja $3,\, 7 \in \mathbb{P}$
