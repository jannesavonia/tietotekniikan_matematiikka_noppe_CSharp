# RSA-salaus

## Harjoituksia

### Harjoituksia

-   **M11.** Toteuta funktio `keys(p, q, e)`, joka palauttaa julkisen
    avaimen $K_{pub}=$`(n, e)` ja yksityisen avaimen
    $K_{priv}=$`(n, d)`. Tässä $0\le n$, $0\le e$ ja $0 \le d$ ovat
    kokonaislukuja.

-   **M12.** Toteuta funktiot `E(Kpub, M)` ja `D(Kpriv, C)`
    RSA-salaukseen. Tässä $0 \le M<n$ ja $0 \le C<n$ ovat
    kokonaislukuja.

-   **13.** Käytännössä lähetettävä viesti joudutaan jakamaan osiin,
    jotta $M<n$ pätee. Jokainen osa salataan erikseen ja vastaanottava
    pää puolestaan purkaa salatut osat ja kokoaa niistä alkuperäisen
    viestin. Toteuta funktiot tekstimuotoisten viestien salaukseen ja
    purkamiseen. $C$ voi olla esimerkiksi `tuple`, joka sisältää salatun
    viestin osat.

-   **14.** RSA-salauksen ja
    [SHA](https://fi.wikipedia.org/wiki/SHA)-tiivisteen avulla voidaan
    toteuttaa myös esimerkiksi [digitaalinen
    allekirjoitus](https://fi.wikipedia.org/wiki/S%C3%A4hk%C3%B6inen_allekirjoitus).
    Tutustu tähän menetelmään esimerkiksi sivuston [RSA and Digital
    Signatures](https://www.geeksforgeeks.org/rsa-and-digital-signatures/)
    kautta.
