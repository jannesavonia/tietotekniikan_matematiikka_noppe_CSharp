# Vektorit

## 2d-koordinaatisto

<hr class="theorem-block-start" style="border:0; border-top:4px solid #1565c0; margin-top:1.2em; margin-bottom:0.6em;">

#### Definition

-   **Tason koordinaatisto** muodostetaan valitsemalla yksi tason
    piste origoksi ja kiinnittämällä siihen kaksi toisiaan vastaan
    kohtisuoraa koordinaattiakselia.

-   Koordinaattiakseleille täytyy kiinnittää **positiiviset suunnat**.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #1565c0; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #00695c; margin-top:1.2em; margin-bottom:0.6em;">

#### Remark

Yleensä käytetään koordinaatistoa, jossa positiiviselta $x$-akselilta
kierryttäessä vastapäivään $90^\circ$ tullaan positiiviselle
$y$-akselille.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #00695c; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #00695c; margin-top:1.2em; margin-bottom:0.6em;">

#### Remark

Jokaisen tason pisteen paikka voidaan ilmaista lukuparina $(x, y)$.
Tässä

-   ensimmäinen luku, $x$, ilmaisee pisteen $x$-koordinaatin ja

-   jälkimmäinen luku, $y$, ilmaisee pisteen $y$-koordinaatin.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #00695c; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #1565c0; margin-top:1.2em; margin-bottom:0.6em;">

#### Definition

Samaistetaan taso $\mathbb{R}^2$:n kanssa:

-   Jokaista tason pistettä vastaa lukupari $(x, y)$ ja

-   jokaista lukuparia $(x, y)$ vastaa yksi tason piste.

-   Merkitään pistettä vastaavaa vektoria jatkossa merkinnällä
    $\begin{pmatrix}x & y\end{pmatrix}$.

#### Remark

Vastaavasti 1d-avaruus voidaan samaistaa $\mathbb{R}$:n (lukusuora)
kanssa.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #00695c; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #1565c0; margin-top:1.2em; margin-bottom:0.6em;">

#### Definition

-   3-ulotteinen avaruus ja $\mathbb{R}^3$ voidaan samaistaa
    vastaavalla tavalla kuin 2d-tapauksessa.

-   Samaistetaan piste $P$ ja $\mathbb{R}^3$:n vektori
    $\begin{pmatrix}P_x & P_y & P_z\end{pmatrix}$, jossa

    -   $P_x$ on pisteen $P$ $x$-koordinaatti,

    -   $P_y$ on pisteen $P$ $y$-koordinaatti ja

    -   $P_z$ on pisteen $P$ $z$-koordinaatti.
