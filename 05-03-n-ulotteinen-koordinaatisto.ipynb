# Vektorit

## n-ulotteinen koordinaatisto

<hr class="theorem-block-start" style="border:0; border-top:4px solid #00695c; margin-top:1.2em; margin-bottom:0.6em;">

#### Remark

-   Vastaavalla tavalla voidaan samaistaa $n$-ulotteinen avaruus ja
    $\mathbb{R}^n$.

-   Kuvan piirtäminen on hankalaa.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #00695c; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #455a64; margin-top:1.2em; margin-bottom:0.6em;">

#### Notation

Muutetaan aiemmin käytettyä merkintää hieman:

-   Numeroidaan akselit, eikä kutsuta niitä enää $x$-, $y$- ja
    $z$-akseleiksi.

-   Merkitään pistettä $x$ vastaavaa $\mathbb{R}^n$:n vektoria
    $\mathbf{x}=\begin{pmatrix}x_1& x_2& \dots& x_n\end{pmatrix}$.

-   Käytetään jatkossa tässä määriteltyä esitystapaa riippumatta
    siitä, mikä avaruuden dimensio on.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #455a64; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #00695c; margin-top:1.2em; margin-bottom:0.6em;">

#### Remark

-   Matemaattisesti vektori on eri asia kuin koordinaatiston piste

-   Vektorin alkupiste voidaan asettaa mihin tahansa koordinaatiston
    pisteeseen ja silti vektori pysyy samana

    -   Vektorin alkupiste ei tarvitse olla origossa

    -   "Vektori voidaan siis siirtää aina haluttuun paikkaan"
