#### Definition

Olkoon $\mathbf{x}, \mathbf{y} \in \mathbb{R}^n$. Tällöin 

$$
\begin{aligned}
            \textbf{x} + \textbf{y} &= \begin{pmatrix}x_1 + y_1 & x_2 + y_2 & \dots & x_n+y_n\end{pmatrix}\text{ ja}\\
            \textbf{x} - \textbf{y} &= \begin{pmatrix}x_1 - y_1 & x_2 - y_2 & \dots & x_n-y_n\end{pmatrix}\ .
\end{aligned}
$$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #1565c0; margin-top:0.6em; margin-bottom:1.2em;">

#### Example 19

$$
\begin{aligned}
            \begin{pmatrix}1 & -1\end{pmatrix} + \begin{pmatrix}2 3\end{pmatrix} & = &  \begin{pmatrix}3 & 2\end{pmatrix}\\
            \begin{pmatrix}0 & 0 1\end{pmatrix} + \begin{pmatrix}0 & -4 & -1\end{pmatrix} & = & \begin{pmatrix}0 & -4 & 0\end{pmatrix}\\
            \begin{pmatrix}\pi & \ln 2 \end{pmatrix} + \begin{pmatrix}\sqrt{2} \ln 1\end{pmatrix} & = & \begin{pmatrix} \pi+\sqrt{2} & \ln 2\end{pmatrix}\\
            \begin{pmatrix}1 & 2\end{pmatrix} + \begin{pmatrix}3 1 & 1\end{pmatrix} &  &  \text{Ei määritelty!}\\
            \begin{pmatrix}1 & -1\end{pmatrix} - \begin{pmatrix}2 3\end{pmatrix} & = &  \begin{pmatrix}-1 & -4\end{pmatrix}\\
\end{aligned}
$$
