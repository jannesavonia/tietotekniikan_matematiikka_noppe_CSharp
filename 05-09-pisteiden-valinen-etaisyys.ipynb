## Pisteiden välinen etäisyys

<hr class="theorem-block-start" style="border:0; border-top:4px solid #ef6c00; margin-top:1.2em; margin-bottom:0.6em;">

#### Example 23

Avaruuden $\mathbb{R}^n$ kahden pisteen $\mathbf{x}$ ja $\mathbf{y}$ etäisyys voidaan
laskea normin avulla.  
Olkoon $\mathbf{x}=\begin{pmatrix}1& 3 & -1 & 1\end{pmatrix}$ ja
$\mathbf{y}=\begin{pmatrix}-1& 3 & 2 & 3\end{pmatrix}$.

Tällöin pisteiden välinen etäisyys 

$$
\begin{aligned}
            \lVert \mathbf{x} - \mathbf{y} \rVert & = \lVert \begin{pmatrix}2 & 0 & -3 & -2\end{pmatrix} \rVert\\
            & = \sqrt{2^2 + 0^2 + (-3)^2 + (-2)^2} \\
            & = \sqrt{4 + 0 + 9 + 4} \\
            & = \sqrt{17}\\
            & = \lVert \mathbf{y} - \mathbf{x} \rVert\ .
\end{aligned}
$$
