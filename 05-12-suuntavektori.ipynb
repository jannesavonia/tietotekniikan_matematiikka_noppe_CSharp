#### Example 26

Monesti tarvitaan tieto kahta avaruuden pistettä yhdistävän janan
suunnasta. Tämä tieto voidaan laskea käyttämällä yksikkövektoria.  
Olkoon $\mathbf{x}=\begin{pmatrix}3 & -1\end{pmatrix}$ ja $\mathbf{y}=\begin{pmatrix}2&  1\end{pmatrix}$. Tällöin
suunta $\mathbf{x}$:stä $\mathbf{y}$:hyn 

$$
\begin{aligned}
            \hat{\mathbf{z}} & = \frac{1}{\lVert \mathbf{y} - \mathbf{x} \rVert}(\mathbf{y} - \mathbf{x})\\
            & = \frac{1}{\lVert \begin{pmatrix}-1 & 2\end{pmatrix} \rVert} \begin{pmatrix}-1 & 2\end{pmatrix} \\
            & = \frac{1}{\sqrt{5}} \begin{pmatrix}-1 & 2\end{pmatrix} \ .
\end{aligned}
$$
