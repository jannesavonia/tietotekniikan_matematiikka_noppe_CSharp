#### Theorem

Olkoon $\mathbf{x}, \mathbf{y} \in \mathbb{R}^n$. Tällöin $\mathbf{x}$ voidaan jakaa
$\mathbf{y}$:n kanssa samansuuntaiseen komponenttiin $P_{\mathbf{y}}\mathbf{x}$ ja
tätä vastaan kohtisuoraan komponenttiin $P_{\perp \mathbf{y}}\mathbf{x}$.  

1.  Jako on yksikäsitteinen,

2.  $P_{\mathbf{y}}\mathbf{x} = (\mathbf{x} \cdot \hat{\mathbf{y}})\hat{\mathbf{y}}=\frac{\mathbf{x} \cdot \mathbf{y}}{\lVert \mathbf{y} \rVert^2} \mathbf{y}$
    ja

3.  $P_{\perp \mathbf{y}}\mathbf{x} = \mathbf{x}-P_{\mathbf{y}}\mathbf{x}$.

Vektoria $P_{\mathbf{y}}\mathbf{x}$ kutsutaan $\mathbf{x}$:n *projektioksi
$\mathbf{y}$:lle*.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #6a1b9a; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #00695c; margin-top:1.2em; margin-bottom:0.6em;">

#### Remark

Edellä siis pätee

1.  $\mathbf{y} \cdot P_{\perp \mathbf{y}}\mathbf{x} = 0$,

2.  $P_{\mathbf{y}}\mathbf{x} \ne \mathbf{0}$ on yhdensuuntainen vektorin $\mathbf{y}$
    kanssa.

3.  $P_{\mathbf{y}}\mathbf{x} + P_{\perp \mathbf{y}}\mathbf{x} = \mathbf{x}$.

#### Remark

Projektio voidaan laske ilman neliöjuurta käyttäen identiteettiä
$$\lVert \mathbf{y} \rVert^2=\mathbf{y} \cdot \mathbf{y}\ ,$$ jolloin
$$P_{\mathbf{y}}\mathbf{x} = \frac{\mathbf{x} \cdot \mathbf{y}}{\mathbf{y} \cdot \mathbf{y}} \mathbf{y}\ .$$

#### Example 34

Olkoon $\mathbf{x}=\begin{pmatrix}1 & -\frac{1}{2}\end{pmatrix}$ ja
$\mathbf{y}=\begin{pmatrix}\frac{2}{3} & \frac{3}{2}\end{pmatrix}$. Tällöin 

$$
\begin{aligned}
            P_{\mathbf{y}}\mathbf{x} & = \frac{\mathbf{x} \cdot \mathbf{y}}{\mathbf{y} \cdot \mathbf{y}} \mathbf{y} \\
            & = \frac{\begin{pmatrix}1 & -\frac{1}{2}\end{pmatrix} \cdot \begin{pmatrix}\frac{2}{3} & \frac{3}{2}\end{pmatrix}}{ \begin{pmatrix}\frac{2}{3} & \frac{3}{2}\end{pmatrix}\cdot \begin{pmatrix}\frac{2}{3} & \frac{3}{2}\end{pmatrix}} \begin{pmatrix}\frac{2}{3} & \frac{3}{2}\end{pmatrix}\\
            & = \frac{\frac{2}{3}-\frac{3}{4}}{\frac{4}{9} + \frac{9}{4}} \begin{pmatrix}\frac{2}{3} & \frac{3}{2}\end{pmatrix} \\
            & = -\frac{9}{291} \begin{pmatrix}\frac{2}{3} & \frac{3}{2}\end{pmatrix} \\
\end{aligned}
$$
