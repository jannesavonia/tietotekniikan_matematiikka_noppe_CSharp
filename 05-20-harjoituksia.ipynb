### Harjoituksia

-   **29.** Olkoon $\mathbf{x}=\begin{pmatrix}-1& 1\end{pmatrix}$ ja $\mathbf{y}=\begin{pmatrix}2& -1\end{pmatrix}$.
    Laske $P_{\mathbf{y}}\mathbf{x}$ ja $P_{\mathbf{x}}\mathbf{y}$.

-   **30.** Olkoon $\mathbf{x}=\begin{pmatrix}-1& 1& 0\end{pmatrix}$ ja
    $\mathbf{y}=\begin{pmatrix}2& -1 & 2\end{pmatrix}$. Laske $P_{\mathbf{y}}\mathbf{x}$ ja
    $P_{\mathbf{x}}\mathbf{y}$.

-   **M18.** Lisää `Vect`-luokkaan metodi `proj(a)`, joka palauttaa
    vektorin projektion `a`:lle. Jos dimensiot eivät täsmää, niin
    heitetään poikkeus.  
    Tämä on viimeinen tehtävä, jossa käytetään `Vect`-luokkaa. Jatkossa
    käytetään NumPy-kirjastoa tai Math.NET Numerics-kirjastoa.
