# Matriisit

## Matriisi

<hr class="theorem-block-start" style="border:0; border-top:4px solid #455a64; margin-top:1.2em; margin-bottom:0.6em;">

#### Notation

Edellä vektorit kirjoitettiin ladontateknisistä syistä johtuen aina
vaakavektoreina. Jatkossa vektori on kuitenkin aina pystyvektori, jos
ei toisin mainita. Vektorien laskusäännöt pysyvät kuitenkin samoina.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #455a64; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #ef6c00; margin-top:1.2em; margin-bottom:0.6em;">

#### Example 35

Aikaisemmin $\mathbf{x}=\begin{pmatrix}x_1&x_2 & \dots & x_n\end{pmatrix}$, jatkossa
$\mathbf{x}=\begin{pmatrix}x_1\\ x_2\\ \vdots \\ x_n\end{pmatrix}$.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #ef6c00; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #1565c0; margin-top:1.2em; margin-bottom:0.6em;">

#### Definition

**Matriisi** on $m \times n$-matriisi lukutaulukko, jossa on $m$ riviä
ja $n$ saraketta. Tällä kurssilla matriisien alkiot ovat reaalilukuja.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #1565c0; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #455a64; margin-top:1.2em; margin-bottom:0.6em;">

#### Notation

Merkitään jatkossa matriiseja isoilla lihavoiduilla kirjaimilla.
Olkoon $\mathbf{A}$ $m \times n$-matriisi. Tällöin merkitään
$\mathbf{A} \in \mathbb{R}^{m \times n}$.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #455a64; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #ef6c00; margin-top:1.2em; margin-bottom:0.6em;">

#### Example 36

Olkoon 

$$
\textbf{A}=\begin{pmatrix}
            1 & 2 \\
            1 & 4 \\
            2 & 4
        \end{pmatrix} \ .
$$

 Tällöin
$\textbf{A} \in \mathbb{R}^{3 \times 2}$.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #ef6c00; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #455a64; margin-top:1.2em; margin-bottom:0.6em;">

#### Notation

Merkitään matriisin $\mathbf{A}$ yksittäistä alkiota $a_{ij}$:llä (tai
$a_{i,j}$:llä jos pilkku on tarpeen). $a_{ij}$ on matriisin $i$:nnen
rivin $j$:s alkio.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #455a64; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #ef6c00; margin-top:1.2em; margin-bottom:0.6em;">

#### Example 37

Olkoon $A \in \mathbb{R}^{3 \times 2}$. Tällöin

$$
\textbf{A}=\begin{pmatrix}
            a_{1,1} & a_{1,2} \\
            a_{2,1} & a_{2,2} \\
            a_{3,1} & a_{3,2}
        \end{pmatrix} \ .
$$
