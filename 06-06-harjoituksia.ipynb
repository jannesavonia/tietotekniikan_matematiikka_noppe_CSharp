### Harjoituksia

-   **31.** Olkoon $\mathbf{A} = \begin{pmatrix}
            1 & 2 \\
            3 & 4 \\
            5 & 6\end{pmatrix}$, $\mathbf{B} = \begin{pmatrix}
            3 & 2 \\
            1 & 3 \\
            1 & 2\end{pmatrix}$ ja $c=4$. Laske $\textbf{A}+c\textbf{B}$.

-   **32.** Olkoon $\mathbf{A}=\begin{pmatrix}
            0.2 & 1.1\\
            -1.1 & 0.2
        \end{pmatrix}$. Laske $\mathbf{A}^T + \mathbf{A}$.

-   **33.** Olkoon $\mathbf{A}$ kuten edellä. Laske $\mathbf{A}^T - 2\mathbf{A}$.
