# Matriisit

## Laskusääntöjä

<hr class="theorem-block-start" style="border:0; border-top:4px solid #6a1b9a; margin-top:1.2em; margin-bottom:0.6em;">

#### Theorem

Olkoon $\textbf{A}, \textbf{B}$ ja $\textbf{C}$ matriiseja ja $c$
skalaari. Mikäli laskutoimitukset ovat määriteltyjä, niin

$$
\begin{aligned}
            \textbf{A} + \textbf{0} & = \textbf{A} \ ,\\
            \textbf{A} + \textbf{B} & = \textbf{B} + \textbf{A} \ ,\\
            \textbf{A} + ( \textbf{B} + \textbf{C} )& = ( \textbf{A} + \textbf{B} ) + \textbf{C} \ ,\\
            c(\textbf{A} + \textbf{B}) & = c\textbf{A} + c\textbf{B} \ ,\\
            (c\textbf{A})\textbf{B} & = c(\textbf{AB}) = \textbf{A}(c\textbf{B}) \ ,\\
            \textbf{A} \textbf{0} & = \textbf{0} \text{ ja } \textbf{0} \textbf{A}  =  \textbf{0} \ ,\\
            \textbf{A} \textbf{I} & = \textbf{A} \text{ ja } \textbf{I} \textbf{A}  =  \textbf{A} \ ,\\
            \textbf{A} ( \textbf{B} \textbf{C} )& = ( \textbf{A} \textbf{B} ) \textbf{C} \ ,\\
            \textbf{A} ( \textbf{B} + \textbf{C} )& = ( \textbf{A} \textbf{B} ) + ( \textbf{A} \textbf{C} ) \ ,\\
            \textbf{A} & = (\textbf{A}^T)^T \ \textnormal{ja} \\
            (\textbf{A}\textbf{B})^T & = \textbf{B}^T\textbf{A}^T \ .
\end{aligned}
$$
