-   **39.** Tutki ovatko seuraavat matriisit kääntyviä:

$$
\begin{aligned}
            \begin{pmatrix}1 & 2\\ 2 1 \end{pmatrix} & \hspace{0.5cm}\begin{pmatrix}1 & 1 \\ 2 & 2\end{pmatrix} & \hspace{0.5cm}\begin{pmatrix}0 & 1 \\ -1 & 0\end{pmatrix} \\
            \begin{pmatrix}1 & 1\\ 0 0 \end{pmatrix} & \hspace{0.5cm}\begin{pmatrix}-1 & 1 \\ 2 & 2\end{pmatrix} & \hspace{0.5cm}\begin{pmatrix}1 & 0 \\ -1 & 0\end{pmatrix}
    \end{aligned}
$$

-   **40.**$2 \times 2$ determinantin laskemiseksi määritelmän mukaan
    tarvitaan 2 kertolaskua ja $3 \times 3$ determinantin laskemiseksi
    tarvitaan vastaavasti $3 \cdot 3 = 9$ kertolaskua. Arvioi
    determinantin määritelmän avulla, montako kertolaskua tarvitaan
    $4 \times 4$ determinantin laskemiseksi.
