# Matriisit

## Ortogonaalinen matriisi

<hr class="theorem-block-start" style="border:0; border-top:4px solid #1565c0; margin-top:1.2em; margin-bottom:0.6em;">

#### Definition

Matriisin $\textbf{A}\in \mathbb{R}^{n \times n}$ sanotaan olevan
**ortogonaalinen** jos
$$\textbf{A}\textbf{A}^T = \textbf{I} = \textbf{A}^T \textbf{A} \ .$$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #1565c0; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #ad1457; margin-top:1.2em; margin-bottom:0.6em;">

#### Corollary

Mikäli matriisi $\mathbf{A}$ on ortogonaalinen, niin sen käänteismatriisi
$$\textbf{A}^{-1} = \textbf{A}^T \ .$$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #ad1457; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #ef6c00; margin-top:1.2em; margin-bottom:0.6em;">

#### Example 53

Aiemmin todettiin, että kiertomatriisin
$\mathbf{R}(\alpha) = \begin{pmatrix} \cos \alpha & -\sin \alpha \\ \sin \alpha & \cos \alpha\end{pmatrix}$
käänteismatriisi $\mathbf{R}^{-1}(\alpha) = \mathbf{R}(-\alpha)$. Edelleen

$$
\begin{aligned}
        \mathbf{R}(-\alpha) & = \begin{pmatrix} \cos -\alpha & -\sin -\alpha \\ \sin -\alpha & \cos -\alpha\end{pmatrix} \\
        & = \begin{pmatrix} \cos \alpha & \sin \alpha \\ -\sin \alpha & \cos \alpha\end{pmatrix} \\
        & = \mathbf{R}^T(\alpha)\ ,
\end{aligned}
$$

 joten matriisi on ortogonaalinen.
