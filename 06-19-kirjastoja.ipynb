```
#r "nuget: MathNet.Numerics, 5.0.0"
#r "nuget: ScottPlot, 5.1.57"

// Source file: examples/mathnetexample02.cs

using System;
using MathNet.Numerics.LinearAlgebra;

//dotnet add package MathNet.Numerics
//dotnet add package ScottPlot

static void DrawArrow(ScottPlot.Plot plot, Vector<double> xBase, Vector<double> xTip) {
    ScottPlot.Coordinates xBaseCoord = new(xBase[0], xBase[1]);
    ScottPlot.Coordinates xTipCoord = new(xTip[0], xTip[1]);
    ScottPlot.CoordinateLine arrow = new(xBaseCoord, xTipCoord);
    plot.Add.Arrow(arrow);
}
```

```
Vector<double> x=CreateVector.Dense(
    new double[] {1.0, 3.0});
```

```
Vector<double> y=CreateVector.Dense(
    new double[] {3.2, 5.1});
```

```
double xy=x*y;
Console.WriteLine($"{x}*{y}={xy}");
```

```
Vector<double> O=CreateVector.Dense(
    new double[] {0, 0});
```

```
ScottPlot.Plot myPlot = new();
double[] x_list=new double[] {O[0], x[0]};
double[] y_list=new double[] {O[1], x[1]};
myPlot.Add.Scatter(x_list, y_list);

x_list=new double[] {O[0], y[0]};
y_list=new double[] {O[1], y[1]};
myPlot.Add.Scatter(x_list, y_list);
myPlot.SavePng("mathnetexample02a.png", 800, 600);
```

```
myPlot = new();
Vector<double> x_minus_y=x-y;
```

```
DrawArrow(myPlot, O, x);
myPlot.Add.Text("x", x[0]/2, x[1]/2);
DrawArrow(myPlot, O, y);
myPlot.Add.Text("y", y[0]/2, y[1]/2);
DrawArrow(myPlot, y, x);
myPlot.Add.Text("x-y", y[0]+x_minus_y[0]/2, y[1]+x_minus_y[1]/2);
myPlot.SavePng("mathnetexample02b.png", 800, 600);
```

```
myPlot = new();
Vector<double> minus_y=-y;
```

```
DrawArrow(myPlot, O, x);
myPlot.Add.Text("x", x[0]/2, x[1]/2);
DrawArrow(myPlot, x, x+minus_y);
myPlot.Add.Text("-y", x[0]+minus_y[0]/2, x[1]+minus_y[1]/2);
DrawArrow(myPlot, O, x+minus_y);
myPlot.Add.Text("x-y", (x[0]+minus_y[0])/2, (x[1]+minus_y[1])/2);
myPlot.SavePng("mathnetexample02c.png", 800, 600);
```

```
#r "nuget: MathNet.Numerics, 5.0.0"
#r "nuget: ScottPlot, 5.1.57"

// Source file: examples/mathnetexample03.cs

using System;
using MathNet.Numerics.LinearAlgebra;

//dotnet add package MathNet.Numerics
//dotnet add package ScottPlot

static void DrawXY(ScottPlot.Plot plot, Vector<double> x, Vector<double> y) {
    double[] x_list=x.ToArray();
    double[] y_list=y.ToArray();
    plot.Add.Scatter(x_list, y_list);
}
```

```
//Insert points into list (forms matrix)
Matrix<double> x = CreateMatrix.DenseOfArray(new double[,] {
    {0, 0}, 
    {2, 1}, 
    {2, 2}, 
    {1, 2}, 
    {0, 0},
    {1, 1},
    {2, 2}
}).Transpose();
Console.WriteLine(x);
```

```
//Create linear mapping (matrix)
Matrix<double> A = CreateMatrix.DenseOfArray(new double[,] {
    {2, 0},
    {0, -1}
});
```

```
var Ax=A*x;
```

```
ScottPlot.Plot myPlot = new();
DrawXY(myPlot, x.Row(0), x.Row(1));
DrawXY(myPlot, Ax.Row(0), Ax.Row(1));

myPlot.SavePng("mathnetexample03.png", 800, 600);
```
