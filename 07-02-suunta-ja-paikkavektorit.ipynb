# Homogeeninen koordinaatisto

## Suunta- ja paikkavektorit

### Suuntavektorit ja paikkavektorit

-   Avaruuden piste esitetään usein **paikkavektorin** avulla

-   Useissa sovelluksissa

    > *esimerkiksi valaistusmallit*

    käytetään **suuntavektoreita**

-   Suuntavektori lasketaan kahden pisteen paikan, eli paikkavektoreiden
    erotuksena

-   Homogeenisessa koordinaatistossa

    -   paikkavektorin viimeinen alkio on $1$

    -   suuntavektorin viimeinen alkio on $0$, koska vähennyslaskussa
        $1$:t kumoavat toisensa

-   Esimerkiksi
    $$\begin{pmatrix}3\\-2\\1\end{pmatrix} - \begin{pmatrix}2\\ 1 \\ 1\end{pmatrix} = \begin{pmatrix}1 \\ -3 \\ 0\end{pmatrix}$$

-   Usein suuntavektori muutetaan yksikkövektoriksi

### Suunta- ja paikkavektoreiden väliset operaatiot

1.  Kahta paikkavektoria ei pidä laskea yhteen

    -   Ei geometrista merkitystä, tulos riippuu origon paikasta
        (=koordinaatiston sijainnista)

    -   Viimeinen koordinaatti saa arvon $2$ $\Rightarrow$ Siirto ei
        toimi

2.  Paikkavektoria ei pidä kertoa skalaarilla, mutta suuntavektorin voi

    -   Viimeisen koordinaatin toiminta

3.  Paikkavektorin ja (skaalatun) suuntavektorin yhteenlasku on sallittu

    -   Tuloksena alkuperäisestä pisteestä haluttuun suuntaan tietyllä
        etäisyydellä sijaitseva piste (paikkavektori)

    -   Viimeinen koordinaatti saa arvon $1$ $\Rightarrow$ Paikkavektori

4.  Paikkavektoreiden pistetulo ei ole sallittu

    -   Koordinaatiston valinta vaikuttaa tulokseen

    -   Viimeinen $1$ vääristää pistetulon tulosta

5.  Suuntavektoreiden välinen pistetulo on sallittu

    -   Suuntien välinen kulma

    -   Viimeisten koordinaattien tulo on $0$

### Ristitulo

-   Usein tarvitaan kahden (suunta-)vektorin määräämän tason normaalia
    $\mathbb{R}^3$:ssa

-   Katso [Youtube - Cross product
    introduction](https://youtu.be/pJzmiywagfY?si=BoXMGjDrs5xmtm9Q)

-   Tässä vektori ei ole homogeenisessa koordinaatistossa

<hr class="theorem-block-start" style="border:0; border-top:4px solid #1565c0; margin-top:1.2em; margin-bottom:0.6em;">

#### Definition

Olkoon $\mathbf{x}$ ja $\mathbf{y} \in \mathbb{R}^3$. Tällöin niiden **ristitulo**

$$
\mathbf{x} \times \mathbf{y} = \begin{pmatrix}
            x_2 y_3 - x_3 y_2\\
            x_3 y_1 - x_1 y_3\\
            x_1 y_2 - x_2 y_1
        \end{pmatrix}
$$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #1565c0; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #ad1457; margin-top:1.2em; margin-bottom:0.6em;">

#### Corollary

1.  Jos $\mathbf{z}=\mathbf{x} \times \mathbf{y}$, niin $\mathbf{x} \cdot \mathbf{z} = 0$ ja
    $\mathbf{y} \cdot \mathbf{z} = 0$

2.  $\mathbf{x} \times \mathbf{y} = -\mathbf{y} \times \mathbf{x}$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #ad1457; margin-top:0.6em; margin-bottom:1.2em;">

-   Ristitulon avulla saadaan siis laskettua pinnan normaali

````
#r "nuget: MathNet.Numerics, 5.0.0"
#r "nuget: ScottPlot, 5.1.57"

// Source file: examples/hgexample01.cs

using System;
using MathNet.Numerics.LinearAlgebra;

static Vector<double> VectorAppend(Vector<double> v, double value) {
      double[] u_list=new double[v.Count+1];
      for(int i=0; i<v.Count; i++) {
            u_list[i] = v[i];
      }
      u_list[v.Count]=value;

      return CreateVector.Dense(u_list);
}
````

````
Vector<double> x_R2=CreateVector.Dense(new double[] {0, 2});
Vector<double> y_R2=CreateVector.Dense(new double[] {-3, 4});

Console.WriteLine($"x_R2={x_R2}");
Console.WriteLine($"y_R2={y_R2}");
````

````
var x=VectorAppend(x_R2, 1);
var y=VectorAppend(y_R2, 1);
Console.WriteLine($"x={x}");
Console.WriteLine($"y={y}");
````

````
var x_to_y=y-x;
Console.WriteLine($"x_to_y={x_to_y}");
````

````
var unit_x_to_y=x_to_y.Normalize(2);
Console.WriteLine($"unit_x_to_y={unit_x_to_y}");
````

``` PseudoPython
# Source file: examples/vectplot.py

import numpy as np 
from matplotlib import pyplot as plt 
import matplotlib.patches as mpatches

def vect3d_plot(x, y, z, x_label, y_label, z_label):
    ax = plt.figure().add_subplot(projection='3d')
    ax.set_aspect('equal')
    ax.axes.set_xlim3d(left=-0.5, right=1.5) 
    ax.axes.set_ylim3d(bottom=-0.5, top=1.5) 
    ax.axes.set_zlim3d(bottom=0, top=2) 
    ax.xaxis.set_ticklabels([])
    ax.yaxis.set_ticklabels([])
    ax.zaxis.set_ticklabels([])
    ax.set_xlabel(r'$x_1$')
    ax.set_ylabel(r'$x_2$')
    ax.set_zlabel(r'$x_3$')

    #Draw projections 1st
    x_proj=np.append(x[:2], [0])
    y_proj=np.append(y[:2], [0])
    z_proj=np.append(z[:2], [0])
    ax.quiver(0, 0, 0, *x_proj, color='gray')
    ax.quiver(0, 0, 0, *y_proj, color='gray')
    ax.quiver(0, 0, 0, *z_proj, color='gray')

    #Draw vectors
    ax.quiver(0, 0, 0, *x, color='r')
    ax.quiver(0, 0, 0, *y, color='b')
    ax.quiver(0, 0, 0, *z, color='m')

    #Add legend
    x_patch = mpatches.Patch(color='r', label=x_label)
    y_patch = mpatches.Patch(color='b', label=y_label)
    z_patch = mpatches.Patch(color='m', label=z_label)
    plt.legend(handles=[x_patch, y_patch, z_patch], 
        loc='upper left', framealpha=0.5, frameon=True)

    return ax
```
