# Homogeeninen koordinaatisto

## Harjoituksia

### Harjoituksia

-   **M21.** Kirjoita $1$-ulotteisen homogeenisen koordinaatiston
    funktiot

    -   `T(dx)`, joka palauttaa matriisin, joka vastaa siirtoa `dx`:n
        verran sekä

    -   `S(c)`, joka palauttaa matriisin, joka vastaa skaalausta
        vakiolla `c`.

-   **M22.** Kirjoita funktio `unitnormal(a, b)`, jossa `a` ja `b` ovat
    $\mathbb{R}^3$:n vektoreita (ei homogeenisissa koordinaateissa). Funktio
    palauttaa `a`:n ja `b`:n määräämän

    > *virittämän*

    tason yksikkönormaalin, eli yksikkövektorin, joka on molempia
    vektoreita vasten kohtisuorassa.

-   **M23.** Kirjoita funktio `direction(a, b)`, joka saa parametrina
    kaksi homogeenisen koordinaatiston

    > *Ei kiinnitetä dimensiota tässä yhteydessä, koska yleisesti
    > toimiva koodi on helppo toteuttaa*

    paikkavektoria. Funktio palauttaa yksikkösuuntavektorin `a`:sta
    `b`:hen.
