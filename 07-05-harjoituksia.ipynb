# Homogeeninen koordinaatisto

## Harjoituksia

### Harjoituksia

-   **M24.** Kirjoita funktio `rotate_around_point(alpha, point, X)`,
    joka kiertää `alpha`

    > *radiaaneissa*

    :n verran homogeenisia koordinaatteja `X` pisteen `point` ympäri
    vastapäivään. Funktio palauttaa muunnetut pisteet matriisissa.
    Kaikki pisteet ovat $\mathbb{R}^2$ pisteitä, jotka on upotettu
    homogeeniseen koordinaatistoon.

-   **M25.** Kappaleen paikka
    $\mathbf{x}(t)=\begin{pmatrix}\cos(\pi t/10) \sin(\pi t) \\ \cos(\pi t)\end{pmatrix}$.
    Lisäksi kappaletta kierretään (kappaleen koordinaatiston origon
    ympäri) vastapäivään $\frac{\pi t}{16}$ radiaania. Kirjoita funktio
    `move_rotate(t)`, joka palauttaa tarvittavan muunnosmatriisin.
