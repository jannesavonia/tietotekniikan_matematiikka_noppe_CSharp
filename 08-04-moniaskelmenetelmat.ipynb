# Differentiaaliyhtälöiden numeerinen ratkaiseminen

## Moniaskelmenetelmät

<hr class="theorem-block-start" style="border:0; border-top:4px solid #0277bd; margin-top:1.2em; margin-bottom:0.6em;">

#### Algorithm

Valitsemalla kertaluku $k$, saadaan eksplisiittisiä
Adams-Bashforth-menetelmiä: 

$$
\begin{aligned}
            k=0 \ : \ & \mathbf{y}_{n+1} = \mathbf{y}_n + \Delta t\, \mathbf{f}_n\\
            k=1 \ : \  & \mathbf{y}_{n+1} = \mathbf{y}_n + \frac{\Delta t}{2} (3\mathbf{f}_n-\mathbf{f}_{n-1})\\
            k=2 \ : \  & \mathbf{y}_{n+1} = \mathbf{y}_n + \frac{\Delta t}{12}(23 \mathbf{f}_n -16 \mathbf{f}_{n-1} + 5\mathbf{f}_{n-2})\\
\end{aligned}
$$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #0277bd; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #00695c; margin-top:1.2em; margin-bottom:0.6em;">

#### Remark

Moniaskelmenetelmät pitää käynnistää jollain yksinkertaisemmalla
menetelmällä, esimerkiksi Eulerin menetelmällä. Tämä tarkoittaa sitä,
että ensimmäiset askeleet lasketaan jollain yksiaskelmenetelmällä,
kunnes riittävän monta $\mathbf{y}$:n arvoa on käytössä, ja vasta sitten
siirrytään käyttämään moniaskelmenetelmää.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #00695c; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #00695c; margin-top:1.2em; margin-bottom:0.6em;">

#### Remark

Yleisesti moniaskelmenetelmät ovat tarkempia kuin esimerkiksi Eulerin
menetelmä.

#### Example 63

Tarkastellaan kappaleen liikettä 2d-koordinaatistossa ajan funktiona.
Tiedetään, että kiihtyvyys $\mathbf{a}(t)=\begin{pmatrix}-\cos t\\ -\sin t\end{pmatrix}$ ja
alkunopeus, hetkellä $t=0$, on $\mathbf{v_0} = \begin{pmatrix}1 \\ 0\end{pmatrix}$.  
Tilannetta voidaan mallintaa DY:llä
$$\mathbf{v}'(t) = \begin{pmatrix}-\cos t\\ -\sin t\end{pmatrix}\ .$$ Valitaan
$\Delta t = 0.1$ ja kirjoitetaan AB-ratkaisija.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #ef6c00; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="code-block-start" style="border:0; border-top:4px solid #455a64; margin-top:1.2em; margin-bottom:0.6em;">

```
#r "nuget: MathNet.Numerics, 5.0.0"
#r "nuget: ScottPlot, 5.1.57"

// Source file: examples/abexample01.cs

using System;
using MathNet.Numerics.LinearAlgebra;
using ScottPlot;

static void DrawArrow(ScottPlot.Plot plot, Vector<double> xBase, Vector<double> xTip) {
    ScottPlot.Coordinates xBaseCoord = new(xBase[0], xBase[1]);
    ScottPlot.Coordinates xTipCoord = new(xTip[0], xTip[1]);
    ScottPlot.CoordinateLine arrow = new(xBaseCoord, xTipCoord);
    plot.Add.Arrow(arrow);
}

static Vector<double> f(double t, Vector<double> v) {
    return CreateVector.Dense(new double[] {-Math.Cos(t), -Math.Sin(t)});
}
```

```
Vector<double> v0=CreateVector.Dense(new double[] {1, 0});
double Dt=0.1;
double t0=0;

Vector<double> v=v0;
double t=t0;
int N_steps=40;
List<double> t_list=new List<double>();
t_list.Add(0);
List<Vector<double>> v_list=new List<Vector<double>>();
v_list.Add(v0);
```

```
List<Vector<double>> f_list=new List<Vector<double>>();
for(int i=1; i<=N_steps; i++) {
    f_list.Add(f(t, v));
    if(f_list.Count<3) {
        v=v+Dt*f_list.Last();
    } else {
        v=v+(Dt/12)*(23*f_list[2]-16*f_list[1]+5*f_list[0]);
        f_list.RemoveAt(0);
    }
    t=t+Dt;
    t_list.Add(t);
    v_list.Add(v);
}
```

```
Vector<double> O=CreateVector.Dense(2, (double)0);
ScottPlot.Plot myPlot = new();
for(int i=0; i<=N_steps; i++) {
    DrawArrow(myPlot, O, v_list[i]);
}

myPlot.SavePng("abexample01.png", 800, 600);
```
