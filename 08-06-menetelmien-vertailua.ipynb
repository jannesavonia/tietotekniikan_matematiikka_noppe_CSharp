# Differentiaaliyhtälöiden numeerinen ratkaiseminen

## Menetelmien vertailua

### 2-kl lineaarinen yhtälö

-   Olkoon $y\textnormal{''}(t) + y\textnormal{'}(t)+ y(t) = 0$
    alkuehdolla $y(0)=1$ ja $y\textnormal{'}(0)=0$

-   Ratkaisu

    > *[MathDF](https://mathdf.com/dif/)*

    on
    $y(t)= e^{-\frac{1}{2}t} (\cos (\frac{\sqrt{3}t}2) + \frac{1}{\sqrt{3}}\sin (\frac{\sqrt{3}t}2))$

-   Numeerisiin ratkaisuihin lisätään apumuuttuja
    $v(t)=y\textnormal{'}(t)$

-   Sijoitetaan tämä alkuperäiseen yhtälöön: 

$$
\begin{aligned}
                0 & = y\textnormal{''}(t) + y\textnormal{'}(t)+ y(t) \\
                & = v\textnormal{'}(t) + v(t)+ y(t) \\
                \Rightarrow v\textnormal{'}(t) & = - v(t) - y(t)
    \end{aligned}
$$

-   Tällöin yhtälö saa muodon
    $$\begin{pmatrix}y\textnormal{'}(t) \\ v\textnormal{'}(t)\end{pmatrix} = \begin{pmatrix}v(t) \\ -v(t) - y(t)\end{pmatrix}$$

```
#!pythonkernel

%reset -f

# Source file: examples/compare01.py

import numpy as np 
from matplotlib import pyplot as plt 

def y(t):
    return np.exp(-0.5*t)*(np.cos((np.sqrt(3)/2)*t)+(1/np.sqrt(3))*np.sin((np.sqrt(3)/2)*t))
```

```
#!pythonkernel

def f(t, z):
    y=z[0]
    v=z[1]
    return np.array([v, -v-y])
```

```
#!pythonkernel

N_steps=50
t_values=np.linspace(0.0, 10.0, num=N_steps+1)
Dt=(t_values[1]-t_values[0])
y_exact=y(t_values)

z=np.array([1, 0])
y_euler=[z[0]]
for prev_t in t_values[0:-1]:
    z=z+Dt*f(prev_t, z)
    y_euler.append(z[0])
```

```
#!pythonkernel

z=np.array([1, 0])
y_ab=[z[0]]
f_list=[f(0, z)]
```

```
#!pythonkernel

for prev_t in t_values[0:-1]:
    f_list.append(f(prev_t, z))
    if len(f_list)<3:
        z=z+Dt*f_list[-1]
    else:
        z=z+(Dt/12)*(23*f_list[-1]-16*f_list[-2]+5*f_list[-3])
        f_list.pop(0)
    y_ab.append(z[0])
```

```
#!pythonkernel

plt.plot(t_values, y_euler, label='Euler')
plt.plot(t_values, y_ab, label='AB')
plt.plot(t_values, y_exact, 'k:', label='exact')
plt.legend()
plt.show()

plt.title('Error')
plt.plot(t_values, np.abs(y_euler-y_exact), label='Euler')
plt.plot(t_values, np.abs(y_ab-y_exact), label='AB')
plt.yscale('log')
plt.legend()
plt.show()
```

### Ei-vakiokertoiminen DY

-   Olkoon $y(0)=1$ ja $y\textnormal{'}(t)=y(t)\sin^2 t$.

-   Jaetaan $y$:llä ja integroidaan puolittain: 

$$
\begin{aligned}
                \int \frac{y\textnormal{'}(t)}{y(t)}\, dt & = \int \sin^2 t\, dt\\
                \Rightarrow\ln \lvert y(t) \rvert + C & = \int \frac{1}{2} - \frac{1}{2} \cos 2t\, dt
                =  \frac{1}{2} (t- \frac{1}{2} \sin 2t) \\
                \Rightarrow \lvert y(t) \rvert &= \exp(\frac{1}{2} (t- \frac{1}{2} \sin 2t)-C) \\  \Rightarrow y(t) &=& \exp(\frac{1}{2} (t- \frac{1}{2} \sin 2t)-C) \\
                \Rightarrow 1 &= y(0)  =  \exp(\frac{1}{2} (0- \frac{1}{2} \sin 2\cdot0)-C) \\
                \Rightarrow 0 & = \frac{1}{2} (0- \frac{1}{2} \sin 0)-C \ \
                \Rightarrow C = 0\ .
    \end{aligned}
$$

-   Tarkka ratkaisu on siis
    $y(t) = \exp (\frac{1}{2}t- \frac{1}{4} \sin 2t)$ .

```
#r "nuget: MathNet.Numerics, 5.0.0"
#r "nuget: ScottPlot, 5.1.57"

// Source file: examples/compare02.cs

using System;
using MathNet.Numerics.LinearAlgebra;
using ScottPlot;

static void DrawXY(ScottPlot.Plot plot, Vector<double> xx, Vector<double> yy, string legend) {
    double[] x_list=xx.ToArray();
    double[] y_list=yy.ToArray();
    var scatter=plot.Add.Scatter(x_list, y_list);
    scatter.LegendText=legend;
}
```

```
static double y(double t) {
    return Math.Exp(0.5*(t-0.5*Math.Sin(2*t)));
}
```

```
static double f(double t, double y) {
    return y*Math.Sin(t)*Math.Sin(t);
}
```

```
int N_steps=100;
Vector<double> t_values=CreateVector.Dense<double>(N_steps+1);
Vector<double> y_exact=CreateVector.Dense<double>(N_steps+1);
for(int i=0; i<t_values.Count; i++) {
    t_values[i]=30.0*(((double)i)/N_steps);
    y_exact[i]=y(t_values[i]);
}
double Dt=(t_values[1]-t_values[0]);
```

```
double yt=1;
Vector<double> y_euler=CreateVector.Dense<double>(N_steps+1);
y_euler[0]=yt;
for(int i=1; i<t_values.Count; i++) {
    //We use the previous value of t to compute f(t,y)
    double t_prev=t_values[i-1];
    yt=yt+Dt*f(t_prev, yt);
    y_euler[i]=yt;
}
```

```
yt=1;
List<double> f_list=new List<double>();
Vector<double> y_ab=CreateVector.Dense<double>(N_steps+1);
y_ab[0]=yt;
for(int i=1; i<t_values.Count; i++) {
    double t_prev=t_values[i-1];
    f_list.Add(f(t_prev, yt));
    if(f_list.Count<3) {
        yt=yt+Dt*f_list.Last();
    } else {
        yt=yt+(Dt/12)*(23*f_list[2]-16*f_list[1]+5*f_list[0]);
        f_list.RemoveAt(0);
    }
    y_ab[i]=yt;
}
```

```
ScottPlot.Plot myPlot = new();
DrawXY(myPlot, t_values, y_euler, "Euler");
DrawXY(myPlot, t_values, y_ab, "Adams-Bashforth");
DrawXY(myPlot, t_values, y_exact, "Exact solution");
myPlot.Legend.Alignment = Alignment.UpperLeft;
myPlot.ShowLegend();
myPlot.SavePng("compare02.png", 800, 600);
```
